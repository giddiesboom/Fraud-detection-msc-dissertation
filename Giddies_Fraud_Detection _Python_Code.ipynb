```python
# ============================================================================
#  MACHINE LEARNING-BASED FRAUD DETECTION IN FINANCIAL TRANSACTIONS
#  Implementation for Chapter 4 (follows the Chapter 3 methodology)
#
#  Comparative study:  3 imbalance treatments  x  2 learning paradigms
#  Supervised:   Logistic Regression, Random Forest, XGBoost
#  Unsupervised: Isolation Forest, Autoencoder
#  Evaluation:   precision, recall, F1, ROC-AUC, PR-AUC, confusion matrix



# %% ==========================================================================
# CELL 1 - Install dependencies and import libraries
# ============================================================================
!pip -q install imbalanced-learn==0.12.3 xgboost shap 2>/dev/null

import os, warnings, time, json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from math import radians
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, IsolationForest
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV
from sklearn.inspection import permutation_importance
from sklearn.metrics import (precision_score, recall_score, f1_score,
                             roc_auc_score, average_precision_score,
                             precision_recall_curve, roc_curve,
                             confusion_matrix, classification_report)
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler
from imblearn.pipeline import Pipeline as ImbPipeline
from xgboost import XGBClassifier

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")
sns.set_style("whitegrid")
plt.rcParams["figure.dpi"] = 120
plt.rcParams["font.size"] = 10

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)

# ---- CONFIG (adjust if runtime/memory is tight) ----------------------------
TRAIN_PATH   = "/content/fraudTrain.csv"
TEST_PATH    = "/content/fraudTest.csv"
OUT_DIR      = "/content/outputs"
FIG_DIR      = f"{OUT_DIR}/figures"
TAB_DIR      = f"{OUT_DIR}/tables"
CV_FOLDS     = 5           # stratified k-fold for the stability analysis
CV_SAMPLE    = 200_000     # subsample size for the CV stage (None = full data)
IF_FIT_SAMPLE= 200_000     # subsample used to fit the Isolation Forest
RUN_SHAP     = True        # set False to skip SHAP (saves a few minutes)
USE_GPU_XGB  = True        # XGBoost on GPU if the Colab runtime has one

for d in (OUT_DIR, FIG_DIR, TAB_DIR):
    os.makedirs(d, exist_ok=True)

def savefig(name):
    plt.tight_layout()
    plt.savefig(f"{FIG_DIR}/{name}.png", bbox_inches="tight", dpi=150)
    plt.show()

def savetable(df, name, index=True):
    df.to_csv(f"{TAB_DIR}/{name}.csv", index=index)
    return df

print("TensorFlow:", tf.__version__, "| GPU:", tf.config.list_physical_devices('GPU'))


# %% ==========================================================================
# CELL 2 - Load data (temporal partitions kept as the train/test hold-out)
# ============================================================================
DATE_COLS = ["trans_date_trans_time", "dob"]
train_raw = pd.read_csv(TRAIN_PATH, parse_dates=DATE_COLS).drop(columns=["Unnamed: 0"])
test_raw  = pd.read_csv(TEST_PATH,  parse_dates=DATE_COLS).drop(columns=["Unnamed: 0"])

print("Train:", train_raw.shape, "| Test:", test_raw.shape)
overview = pd.DataFrame({
    "Partition": ["Train (fraudTrain)", "Test (fraudTest)", "Total"],
    "Transactions": [len(train_raw), len(test_raw), len(train_raw)+len(test_raw)],
    "Fraud cases": [int(train_raw.is_fraud.sum()), int(test_raw.is_fraud.sum()),
                    int(train_raw.is_fraud.sum()+test_raw.is_fraud.sum())],
})
overview["Fraud %"] = (100*overview["Fraud cases"]/overview["Transactions"]).round(3)
savetable(overview, "table_4_1_dataset_overview", index=False)
print(overview.to_string(index=False))


# %% ==========================================================================
# CELL 3 - Exploratory Data Analysis (figures + tables)
# ============================================================================
df_eda = train_raw.copy()
df_eda["hour"] = df_eda.trans_date_trans_time.dt.hour

# Fig 4.1 - class distribution (log scale)
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
for a, d, t in zip(ax, [train_raw, test_raw], ["Train", "Test"]):
    vc = d.is_fraud.value_counts().sort_index()
    a.bar(["Legitimate (0)", "Fraud (1)"], vc.values, color=["#3E6D9C", "#C0392B"])
    a.set_yscale("log"); a.set_title(f"{t} class distribution (log scale)")
    for i, v in enumerate(vc.values):
        a.text(i, v, f"{v:,}", ha="center", va="bottom")
savefig("fig_4_1_class_distribution")

# Fig 4.2 - fraud rate by spending category
cat = (train_raw.groupby("category").is_fraud.mean().mul(100)
       .sort_values(ascending=False))
plt.figure(figsize=(10, 4.5))
sns.barplot(x=cat.values, y=cat.index, color="#3E6D9C")
plt.xlabel("Fraud rate (%)"); plt.ylabel(""); plt.title("Fraud rate by spending category")
savefig("fig_4_2_fraud_rate_by_category")
savetable(cat.round(3).to_frame("fraud_rate_%"), "table_4_2_fraud_rate_by_category")

# Fig 4.3 - fraud rate by hour of day
hr = df_eda.groupby("hour").is_fraud.mean().mul(100)
plt.figure(figsize=(9, 4))
plt.plot(hr.index, hr.values, marker="o", color="#C0392B")
plt.xlabel("Hour of day"); plt.ylabel("Fraud rate (%)")
plt.title("Fraud rate by hour of day"); plt.xticks(range(0, 24))
savefig("fig_4_3_fraud_rate_by_hour")

# Fig 4.4 - transaction amount by class (log scale)
plt.figure(figsize=(8, 4.5))
for lbl, col in [(0, "#3E6D9C"), (1, "#C0392B")]:
    sns.kdeplot(np.log1p(train_raw.loc[train_raw.is_fraud == lbl, "amt"]),
                fill=True, alpha=.4, color=col,
                label=("Fraud" if lbl else "Legitimate"))
plt.xlabel("log(1 + amount)"); plt.title("Transaction amount distribution by class")
plt.legend()
savefig("fig_4_4_amount_by_class")

# Fig 4.5 - amount summary boxplot by class
plt.figure(figsize=(7, 4))
sns.boxplot(data=train_raw, x="is_fraud", y="amt", showfliers=False,
            palette=["#3E6D9C", "#C0392B"])
plt.yscale("log"); plt.xlabel("is_fraud"); plt.ylabel("amount (log)")
plt.title("Transaction amount by class")
savefig("fig_4_5_amount_boxplot")

# amount summary table
amt_tab = train_raw.groupby("is_fraud").amt.agg(["mean", "median", "std", "max"]).round(2)
savetable(amt_tab, "table_4_3_amount_summary")
print(amt_tab)


# %% ==========================================================================
# CELL 4 - Feature engineering  (Objective 2)  [leakage-safe]
#   - decompose timestamp, derive age, haversine distance
#   - per-card behavioural aggregates (causal / past-only)
#   - frequency encoding fit on TRAIN only; one-hot for low-cardinality
# ============================================================================
def haversine(lat1, lon1, lat2, lon2):
    R = 6371.0
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat, dlon = lat2 - lat1, lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1)*np.cos(lat2)*np.sin(dlon/2)**2
    return R * 2 * np.arcsin(np.sqrt(a))

def add_causal_features(df):
    """Per-card behavioural features using only each card's PAST transactions."""
    df = df.sort_values(["cc_num", "trans_date_trans_time"]).reset_index(drop=True)
    g = df.groupby("cc_num")
    # mean of the card's previous transaction amounts (shifted -> no leakage)
    cum_sum = g["amt"].cumsum() - df["amt"]
    cum_cnt = g.cumcount()
    df["amt_prev_mean"] = (cum_sum / cum_cnt.replace(0, np.nan))
    df["amt_ratio"]     = df["amt"] / df["amt_prev_mean"]
    # seconds since the card's previous transaction
    df["time_since_prev"] = g["unix_time"].diff()
    # number of earlier transactions by the same card on the same calendar day
    df["date"] = df["trans_date_trans_time"].dt.date
    df["txn_today"] = df.groupby(["cc_num", "date"]).cumcount()
    return df

def build_features(df, freq_maps=None, fit=False, medians=None):
    df = add_causal_features(df)
    # time features
    dt = df["trans_date_trans_time"]
    df["hour"]      = dt.dt.hour
    df["dayofweek"] = dt.dt.dayofweek
    df["month"]     = dt.dt.month
    df["is_night"]  = df["hour"].isin([0, 1, 2, 3, 4, 5]).astype(int)
    # customer age at time of transaction
    df["age"] = ((dt - df["dob"]).dt.days / 365.25)
    # log amount + geographic distance
    df["log_amt"]  = np.log1p(df["amt"])
    df["distance_km"] = haversine(df["lat"], df["long"], df["merch_lat"], df["merch_long"])
    # gender to binary
    df["gender_M"] = (df["gender"] == "M").astype(int)

    HIGH_CARD = ["merchant", "job", "city", "state", "zip"]
    if fit:
        freq_maps = {c: df[c].value_counts(normalize=True) for c in HIGH_CARD}
    for c in HIGH_CARD:
        df[f"{c}_freq"] = df[c].map(freq_maps[c]).fillna(0.0)

    # one-hot for the 14-level spending category
    df = pd.get_dummies(df, columns=["category"], prefix="cat")

    # fill behavioural NaNs (first transaction per card) with train medians
    fill_cols = ["amt_prev_mean", "amt_ratio", "time_since_prev"]
    if fit:
        medians = {c: df[c].median() for c in fill_cols}
    for c in fill_cols:
        df[c] = df[c].fillna(medians[c])
    df["amt_ratio"] = df["amt_ratio"].replace([np.inf, -np.inf], medians["amt_ratio"])

    drop = ["trans_date_trans_time", "dob", "cc_num", "merchant", "job", "city",
            "state", "zip", "gender", "first", "last", "street", "trans_num",
            "unix_time", "date", "lat", "long", "merch_lat", "merch_long"]
    df = df.drop(columns=[c for c in drop if c in df.columns])
    return df, freq_maps, medians

train_fe, freq_maps, medians = build_features(train_raw, fit=True)
test_fe,  _, _               = build_features(test_raw, freq_maps=freq_maps,
                                              fit=False, medians=medians)

# align columns (in case a category level is missing in one partition)
y_train = train_fe.pop("is_fraud").astype(int).values
y_test  = test_fe.pop("is_fraud").astype(int).values
train_fe, test_fe = train_fe.align(test_fe, join="left", axis=1, fill_value=0)

FEATURES = list(train_fe.columns)
X_train = train_fe.astype(np.float32).values
X_test  = test_fe.astype(np.float32).values
print(f"Engineered feature matrix: {X_train.shape[1]} features")
print("Features:", FEATURES)
savetable(pd.DataFrame({"feature": FEATURES}), "table_4_4_feature_list", index=False)


# %% ==========================================================================
# CELL 5 - Evaluation helpers
# ============================================================================
def best_f1_threshold(y_true, scores):
    """Threshold that maximises F1 on the precision-recall curve."""
    p, r, thr = precision_recall_curve(y_true, scores)
    f1 = np.divide(2*p*r, p+r, out=np.zeros_like(p), where=(p+r) > 0)
    k = int(np.nanargmax(f1[:-1])) if len(thr) else 0
    return thr[k] if len(thr) else 0.5

def evaluate(name, y_true, scores, threshold=None, family="Supervised", treatment="-"):
    if threshold is None:
        threshold = best_f1_threshold(y_true, scores)
    y_pred = (scores >= threshold).astype(int)
    return {
        "Model": name, "Paradigm": family, "Treatment": treatment,
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "Recall":    recall_score(y_true, y_pred, zero_division=0),
        "F1":        f1_score(y_true, y_pred, zero_division=0),
        "ROC_AUC":   roc_auc_score(y_true, scores),
        "PR_AUC":    average_precision_score(y_true, scores),
        "Threshold": float(threshold),
        "_scores": scores, "_pred": y_pred,
    }

def prior_correct(scores, pi_train, pi_true):
    """Saerens/Dal Pozzolo prior correction after undersampling."""
    num = scores * (pi_true/pi_train)
    den = num + (1-scores)*((1-pi_true)/(1-pi_train))
    return np.clip(num/den, 1e-9, 1-1e-9)

def plot_confusion(y_true, y_pred, title, fname):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(4.2, 3.6))
    sns.heatmap(cm, annot=True, fmt=",d", cmap="Blues", cbar=False,
                xticklabels=["Legit", "Fraud"], yticklabels=["Legit", "Fraud"])
    plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.title(title)
    savefig(fname)


# %% ==========================================================================
# CELL 6 - Supervised experiments  (Objectives 3 & 4)
#   3 models x {baseline, SMOTE, undersampling, cost-sensitive}
#   resampling lives INSIDE the pipeline -> applied to training data only
# ============================================================================
pos = int(y_train.sum()); neg = int(len(y_train)-pos)
spw = neg/pos                      # XGBoost scale_pos_weight
pi_true = pos/len(y_train)         # true fraud prior (for undersampling calibration)

def make_model(kind, cost=False):
    if kind == "LogReg":
        return LogisticRegression(max_iter=1000, n_jobs=-1,
                                  class_weight=("balanced" if cost else None))
    if kind == "RandomForest":
        return RandomForestClassifier(n_estimators=100, max_depth=None,
                                      min_samples_leaf=5, n_jobs=-1,
                                      random_state=RANDOM_STATE,
                                      class_weight=("balanced" if cost else None))
    if kind == "XGBoost":
        return XGBClassifier(n_estimators=400, max_depth=6, learning_rate=0.1,
                             subsample=0.9, colsample_bytree=0.9,
                             eval_metric="aucpr", random_state=RANDOM_STATE,
                             tree_method="hist",
                             device=("cuda" if USE_GPU_XGB else "cpu"),
                             scale_pos_weight=(spw if cost else 1))

def build_pipeline(kind, treatment):
    steps = [("scaler", StandardScaler())]
    if treatment == "SMOTE":
        steps.append(("res", SMOTE(random_state=RANDOM_STATE)))
    elif treatment == "Undersample":
        steps.append(("res", RandomUnderSampler(random_state=RANDOM_STATE)))
    steps.append(("clf", make_model(kind, cost=(treatment == "Cost-sensitive"))))
    return ImbPipeline(steps)

MODELS     = ["LogReg", "RandomForest", "XGBoost"]
TREATMENTS = ["Baseline", "SMOTE", "Undersample", "Cost-sensitive"]

sup_results = []
sup_store   = {}
for kind in MODELS:
    for treat in TREATMENTS:
        t0 = time.time()
        pipe = build_pipeline(kind, treat)
        pipe.fit(X_train, y_train)
        if hasattr(pipe, "predict_proba"):
            scores = pipe.predict_proba(X_test)[:, 1]
        else:
            scores = pipe.decision_function(X_test)
        if treat == "Undersample":                      # recalibrate probabilities
            scores = prior_correct(scores, 0.5, pi_true)
        res = evaluate(kind, y_test, scores, family="Supervised", treatment=treat)
        res["Runtime_s"] = round(time.time()-t0, 1)
        sup_results.append({k: v for k, v in res.items() if not k.startswith("_")})
        sup_store[(kind, treat)] = res
        print(f"{kind:13s} | {treat:14s} | F1={res['F1']:.3f} "
              f"PR-AUC={res['PR_AUC']:.3f} ROC-AUC={res['ROC_AUC']:.3f} "
              f"({res['Runtime_s']}s)")

sup_df = pd.DataFrame(sup_results).round(4)
savetable(sup_df, "table_4_5_supervised_all_results", index=False)
print("\n", sup_df.to_string(index=False))


# %% ==========================================================================
# CELL 7 - Supervised comparison figures
# ============================================================================
# Fig 4.6 - heatmap of PR-AUC across model x treatment
pivot = sup_df.pivot(index="Model", columns="Treatment", values="PR_AUC")[TREATMENTS]
plt.figure(figsize=(7.5, 3.6))
sns.heatmap(pivot, annot=True, fmt=".3f", cmap="YlGnBu")
plt.title("PR-AUC by model and imbalance treatment"); plt.ylabel("")
savefig("fig_4_6_prauc_heatmap")

# Fig 4.7 - grouped bars for F1
plt.figure(figsize=(9, 4.5))
sns.barplot(data=sup_df, x="Model", y="F1", hue="Treatment")
plt.title("F1-score by model and imbalance treatment"); plt.ylim(0, 1)
plt.legend(title="Treatment", bbox_to_anchor=(1.01, 1), loc="upper left")
savefig("fig_4_7_f1_grouped_bar")

# Fig 4.8 - grouped bars for Recall vs Precision trade-off (best treatment/model)
melt = sup_df.melt(id_vars=["Model", "Treatment"],
                   value_vars=["Precision", "Recall", "F1"],
                   var_name="Metric", value_name="Value")
g = sns.catplot(data=melt, x="Treatment", y="Value", hue="Metric", col="Model",
                kind="bar", height=3.6, aspect=0.9)
g.set_xticklabels(rotation=30, ha="right")
plt.savefig(f"{FIG_DIR}/fig_4_8_precision_recall_f1.png", bbox_inches="tight", dpi=150)
plt.show()

# ROC and PR overlays for the best treatment of each model (by PR-AUC)
best_treat = (sup_df.sort_values("PR_AUC", ascending=False)
              .groupby("Model").first()["Treatment"].to_dict())

plt.figure(figsize=(6, 5))
for kind in MODELS:
    r = sup_store[(kind, best_treat[kind])]
    fpr, tpr, _ = roc_curve(y_test, r["_scores"])
    plt.plot(fpr, tpr, label=f"{kind} [{best_treat[kind]}] AUC={r['ROC_AUC']:.3f}")
plt.plot([0, 1], [0, 1], "k--", lw=.8)
plt.xlabel("False positive rate"); plt.ylabel("True positive rate")
plt.title("ROC curves (best treatment per model)"); plt.legend()
savefig("fig_4_9_roc_supervised")

plt.figure(figsize=(6, 5))
for kind in MODELS:
    r = sup_store[(kind, best_treat[kind])]
    p, rc, _ = precision_recall_curve(y_test, r["_scores"])
    plt.plot(rc, p, label=f"{kind} [{best_treat[kind]}] PR-AUC={r['PR_AUC']:.3f}")
plt.axhline(pi_true, ls="--", c="grey", lw=.8, label=f"Baseline={pi_true:.4f}")
plt.xlabel("Recall"); plt.ylabel("Precision")
plt.title("Precision-Recall curves (best treatment per model)"); plt.legend()
savefig("fig_4_10_pr_supervised")

# confusion matrices for each model's best treatment
for kind in MODELS:
    r = sup_store[(kind, best_treat[kind])]
    plot_confusion(y_test, r["_pred"],
                   f"{kind} [{best_treat[kind]}]",
                   f"fig_4_11_cm_{kind}")


# %% ==========================================================================
# CELL 8 - Stratified 5-fold cross-validation (stability, on a subsample)
#   Reports mean +/- std of PR-AUC per model x treatment (Objective 5)
# ============================================================================
if CV_SAMPLE and CV_SAMPLE < len(X_train):
    idx = np.random.RandomState(RANDOM_STATE).choice(len(X_train), CV_SAMPLE, replace=False)
    Xcv, ycv = X_train[idx], y_train[idx]
else:
    Xcv, ycv = X_train, y_train

skf = StratifiedKFold(n_splits=CV_FOLDS, shuffle=True, random_state=RANDOM_STATE)
cv_rows = []
for kind in MODELS:
    for treat in TREATMENTS:
        aps = []
        for tr, va in skf.split(Xcv, ycv):
            pipe = build_pipeline(kind, treat)
            pipe.fit(Xcv[tr], ycv[tr])
            s = pipe.predict_proba(Xcv[va])[:, 1]
            aps.append(average_precision_score(ycv[va], s))
        cv_rows.append({"Model": kind, "Treatment": treat,
                        "PR_AUC_mean": np.mean(aps), "PR_AUC_std": np.std(aps)})
        print(f"{kind:13s} | {treat:14s} | PR-AUC {np.mean(aps):.3f} +/- {np.std(aps):.3f}")

cv_df = pd.DataFrame(cv_rows).round(4)
savetable(cv_df, "table_4_6_cv_stability", index=False)

# Fig 4.12 - CV mean +/- std
plt.figure(figsize=(9, 4.5))
order = cv_df.sort_values("PR_AUC_mean")
lbl = order["Model"] + " | " + order["Treatment"]
plt.errorbar(order["PR_AUC_mean"], range(len(order)),
             xerr=order["PR_AUC_std"], fmt="o", color="#3E6D9C", capsize=3)
plt.yticks(range(len(order)), lbl); plt.xlabel("PR-AUC (5-fold mean +/- std)")
plt.title("Cross-validation stability")
savefig("fig_4_12_cv_stability")


# %% ==========================================================================
# CELL 9 - Hyperparameter tuning of the leading model (RandomizedSearch)
# ============================================================================
param_dist = {
    "clf__n_estimators": [200, 400, 600, 800],
    "clf__max_depth":    [4, 6, 8, 10],
    "clf__learning_rate":[0.03, 0.05, 0.1, 0.2],
    "clf__subsample":    [0.7, 0.8, 0.9, 1.0],
    "clf__colsample_bytree": [0.7, 0.8, 0.9, 1.0],
}
tune_pipe = ImbPipeline([("scaler", StandardScaler()),
                         ("clf", make_model("XGBoost", cost=True))])
search = RandomizedSearchCV(tune_pipe, param_dist, n_iter=15,
                            scoring="average_precision", cv=3,
                            random_state=RANDOM_STATE, n_jobs=-1, verbose=1)
search.fit(Xcv, ycv)
print("Best PR-AUC (CV):", round(search.best_score_, 4))
print("Best params:", search.best_params_)
best_params = {k.replace("clf__", ""): v for k, v in search.best_params_.items()}
savetable(pd.DataFrame([best_params]), "table_4_7_best_xgb_params", index=False)

# refit tuned XGBoost on full training data, evaluate on test
tuned = make_model("XGBoost", cost=True).set_params(**best_params)
sc = StandardScaler().fit(X_train)
tuned.fit(sc.transform(X_train), y_train)
tuned_scores = tuned.predict_proba(sc.transform(X_test))[:, 1]
tuned_res = evaluate("XGBoost (tuned)", y_test, tuned_scores,
                     family="Supervised", treatment="Cost-sensitive")
print(f"Tuned XGBoost -> F1={tuned_res['F1']:.3f} "
      f"PR-AUC={tuned_res['PR_AUC']:.3f} ROC-AUC={tuned_res['ROC_AUC']:.3f}")


# %% ==========================================================================
# CELL 10 - Unsupervised model 1: Isolation Forest
#   label-free operating point: contamination = training fraud rate
# ============================================================================
sc_u = StandardScaler().fit(X_train)
Xtr_s, Xte_s = sc_u.transform(X_train), sc_u.transform(X_test)

if_idx = np.random.RandomState(RANDOM_STATE).choice(
    len(Xtr_s), min(IF_FIT_SAMPLE, len(Xtr_s)), replace=False)
iso = IsolationForest(n_estimators=200, contamination=pi_true,
                      random_state=RANDOM_STATE, n_jobs=-1)
iso.fit(Xtr_s[if_idx])
iso_scores = -iso.score_samples(Xte_s)          # higher = more anomalous
iso_thr = np.quantile(-iso.score_samples(Xtr_s[if_idx]), 1-pi_true)
iso_res = evaluate("Isolation Forest", y_test, iso_scores,
                   threshold=iso_thr, family="Unsupervised", treatment="None")
print(f"Isolation Forest -> F1={iso_res['F1']:.3f} "
      f"PR-AUC={iso_res['PR_AUC']:.3f} ROC-AUC={iso_res['ROC_AUC']:.3f}")
plot_confusion(y_test, iso_res["_pred"], "Isolation Forest", "fig_4_13_cm_isoforest")

# Fig 4.14 - anomaly score distribution by class
plt.figure(figsize=(8, 4))
for lbl, col in [(0, "#3E6D9C"), (1, "#C0392B")]:
    sns.kdeplot(iso_scores[y_test == lbl], fill=True, alpha=.4, color=col,
                label=("Fraud" if lbl else "Legitimate"))
plt.axvline(iso_thr, ls="--", c="k", label="Threshold")
plt.xlabel("Anomaly score"); plt.title("Isolation Forest anomaly scores by class")
plt.legend()
savefig("fig_4_14_isoforest_scores")


# %% ==========================================================================
# CELL 11 - Unsupervised model 2: Autoencoder
#   trained on LEGITIMATE training rows only; threshold from legit error percentile
# ============================================================================
Xtr_legit = Xtr_s[y_train == 0]
val_cut = int(0.9*len(Xtr_legit))
X_ae_tr, X_ae_val = Xtr_legit[:val_cut], Xtr_legit[val_cut:]
n_in = Xtr_s.shape[1]

inp = Input(shape=(n_in,))
e = Dense(32, activation="relu")(inp)
e = Dense(16, activation="relu")(e)
b = Dense(8,  activation="relu")(e)          # bottleneck
d = Dense(16, activation="relu")(b)
d = Dense(32, activation="relu")(d)
out = Dense(n_in, activation="linear")(d)
ae = Model(inp, out); ae.compile(optimizer="adam", loss="mse")
es = EarlyStopping(patience=5, restore_best_weights=True, monitor="val_loss")
hist = ae.fit(X_ae_tr, X_ae_tr, validation_data=(X_ae_val, X_ae_val),
              epochs=50, batch_size=2048, callbacks=[es], verbose=1)

def recon_error(X):
    return np.mean((ae.predict(X, batch_size=4096, verbose=0) - X)**2, axis=1)

ae_scores = recon_error(Xte_s)
ae_thr = np.quantile(recon_error(X_ae_val), 1-pi_true)   # label-free threshold
ae_res = evaluate("Autoencoder", y_test, ae_scores, threshold=ae_thr,
                  family="Unsupervised", treatment="None")
print(f"Autoencoder -> F1={ae_res['F1']:.3f} "
      f"PR-AUC={ae_res['PR_AUC']:.3f} ROC-AUC={ae_res['ROC_AUC']:.3f}")

# Fig 4.15 - training loss curve
plt.figure(figsize=(7, 4))
plt.plot(hist.history["loss"], label="train")
plt.plot(hist.history["val_loss"], label="validation")
plt.xlabel("Epoch"); plt.ylabel("MSE loss"); plt.title("Autoencoder training curve")
plt.legend()
savefig("fig_4_15_autoencoder_loss")

# Fig 4.16 - reconstruction error by class
plt.figure(figsize=(8, 4))
for lbl, col in [(0, "#3E6D9C"), (1, "#C0392B")]:
    sns.kdeplot(np.log1p(ae_scores[y_test == lbl]), fill=True, alpha=.4, color=col,
                label=("Fraud" if lbl else "Legitimate"))
plt.axvline(np.log1p(ae_thr), ls="--", c="k", label="Threshold")
plt.xlabel("log(1 + reconstruction error)")
plt.title("Autoencoder reconstruction error by class"); plt.legend()
savefig("fig_4_16_autoencoder_error")
plot_confusion(y_test, ae_res["_pred"], "Autoencoder", "fig_4_17_cm_autoencoder")


# %% ==========================================================================
# CELL 12 - Feature importance & interpretation  (Objective 5)
# ============================================================================
# gain-based importance from the tuned XGBoost
imp = (pd.Series(tuned.feature_importances_, index=FEATURES)
       .sort_values(ascending=False))
savetable(imp.round(4).to_frame("importance"), "table_4_8_xgb_importance")
plt.figure(figsize=(8, 6))
sns.barplot(x=imp.values[:20], y=imp.index[:20], color="#3E6D9C")
plt.xlabel("Gain importance"); plt.title("XGBoost feature importance (top 20)")
savefig("fig_4_18_xgb_importance")

# random forest importance (best-treatment RF)
rf_best = sup_store[("RandomForest", best_treat["RandomForest"])]
rf_pipe = build_pipeline("RandomForest", best_treat["RandomForest"]).fit(X_train, y_train)
rf_imp = (pd.Series(rf_pipe.named_steps["clf"].feature_importances_, index=FEATURES)
          .sort_values(ascending=False))
plt.figure(figsize=(8, 6))
sns.barplot(x=rf_imp.values[:20], y=rf_imp.index[:20], color="#2E8B57")
plt.xlabel("Gini importance"); plt.title("Random Forest feature importance (top 20)")
savefig("fig_4_19_rf_importance")

# permutation importance on a test subsample (model-agnostic)
perm_idx = np.random.RandomState(RANDOM_STATE).choice(len(X_test),
                                                      min(30000, len(X_test)), replace=False)
perm = permutation_importance(tuned, sc.transform(X_test[perm_idx]), y_test[perm_idx],
                              scoring="average_precision", n_repeats=5,
                              random_state=RANDOM_STATE, n_jobs=-1)
perm_s = pd.Series(perm.importances_mean, index=FEATURES).sort_values(ascending=False)
savetable(perm_s.round(4).to_frame("perm_importance"), "table_4_9_permutation_importance")
plt.figure(figsize=(8, 6))
sns.barplot(x=perm_s.values[:20], y=perm_s.index[:20], color="#C0392B")
plt.xlabel("Drop in PR-AUC when shuffled"); plt.title("Permutation importance (top 20)")
savefig("fig_4_20_permutation_importance")

# optional SHAP summary for the tuned XGBoost
if RUN_SHAP:
    try:
        import shap
        shap_idx = np.random.RandomState(RANDOM_STATE).choice(len(X_test), 3000, replace=False)
        explainer = shap.TreeExplainer(tuned)
        sv = explainer.shap_values(sc.transform(X_test[shap_idx]))
        shap.summary_plot(sv, features=sc.transform(X_test[shap_idx]),
                          feature_names=FEATURES, show=False, max_display=20)
        plt.title("SHAP summary (tuned XGBoost)")
        savefig("fig_4_21_shap_summary")
    except Exception as e:
        print("SHAP skipped:", e)


# %% ==========================================================================
# CELL 13 - Final comparison across all models (best supervised + unsupervised)
# ============================================================================
final_rows, final_store = [], {}
for kind in MODELS:                         # best supervised per model
    r = sup_store[(kind, best_treat[kind])]
    final_store[f"{kind} [{best_treat[kind]}]"] = r
    final_rows.append({k: v for k, v in r.items() if not k.startswith("_")})
for r in (tuned_res, iso_res, ae_res):      # tuned + unsupervised
    final_store[r["Model"]] = r
    final_rows.append({k: v for k, v in r.items() if not k.startswith("_")})

final_df = pd.DataFrame(final_rows)[
    ["Model", "Paradigm", "Treatment", "Precision", "Recall", "F1",
     "ROC_AUC", "PR_AUC", "Threshold"]].round(4)
savetable(final_df, "table_4_10_final_comparison", index=False)
print(final_df.to_string(index=False))

# Fig 4.22 - final PR-AUC / F1 / Recall comparison
fm = final_df.melt(id_vars=["Model", "Paradigm"],
                   value_vars=["PR_AUC", "F1", "Recall"],
                   var_name="Metric", value_name="Value")
plt.figure(figsize=(11, 5))
sns.barplot(data=fm, x="Model", y="Value", hue="Metric")
plt.xticks(rotation=25, ha="right"); plt.title("Final model comparison")
plt.legend(title="Metric", bbox_to_anchor=(1.01, 1), loc="upper left")
savefig("fig_4_22_final_comparison")

# Fig 4.23 - combined PR curves (supervised vs unsupervised)
plt.figure(figsize=(6.5, 5.5))
for name, r in final_store.items():
    p, rc, _ = precision_recall_curve(y_test, r["_scores"])
    ls = "-" if r["Paradigm"] == "Supervised" else "--"
    plt.plot(rc, p, ls, label=f"{name} (PR-AUC={r['PR_AUC']:.3f})")
plt.axhline(pi_true, ls=":", c="grey", label=f"Baseline={pi_true:.4f}")
plt.xlabel("Recall"); plt.ylabel("Precision")
plt.title("Precision-Recall: supervised vs unsupervised"); plt.legend(fontsize=8)
savefig("fig_4_23_final_pr_curves")

# Fig 4.24 - combined ROC curves
plt.figure(figsize=(6.5, 5.5))
for name, r in final_store.items():
    fpr, tpr, _ = roc_curve(y_test, r["_scores"])
    ls = "-" if r["Paradigm"] == "Supervised" else "--"
    plt.plot(fpr, tpr, ls, label=f"{name} (AUC={r['ROC_AUC']:.3f})")
plt.plot([0, 1], [0, 1], "k:", lw=.8)
plt.xlabel("False positive rate"); plt.ylabel("True positive rate")
plt.title("ROC: supervised vs unsupervised"); plt.legend(fontsize=8)
savefig("fig_4_24_final_roc_curves")


# %% ==========================================================================
# CELL 14 - Zip all outputs for download
# ============================================================================
import shutil
shutil.make_archive("/content/fraud_outputs", "zip", OUT_DIR)
print("\nSaved figures:", len(os.listdir(FIG_DIR)), "| tables:", len(os.listdir(TAB_DIR)))
print("Download: /content/fraud_outputs.zip")
try:
    from google.colab import files
    files.download("/content/fraud_outputs.zip")
except Exception:
    pass
```
